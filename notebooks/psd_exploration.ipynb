# PSD Exploration

Quick notebook to inspect captured PSD libraries stored in `data/sample_results/`.

```python
from pathlib import Path
import torch
import numpy as np
import matplotlib.pyplot as plt

repo_root = Path.cwd().resolve().parent
artifact_path = repo_root / 'data' / 'sample_results' / 'psd_log_BPSK.pth'
artifact = torch.load(artifact_path, map_location='cpu', weights_only=False)
artifact.keys()
```

```python
snrs = np.array(artifact['snrs'])
psd = np.asarray(artifact['psds'][0]).reshape(-1)

print('samples:', len(artifact['psds']))
print('snr min/max/mean:', snrs.min(), snrs.max(), snrs.mean())

plt.figure(figsize=(10, 4))
plt.plot(psd)
plt.title('Example PSD sample')
plt.xlabel('Frequency bin')
plt.ylabel('PSD (dB)')
plt.tight_layout()
```
